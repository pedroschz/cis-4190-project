```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
df = pd.read_csv('data/processed/headlines.csv')
print('rows:', len(df))
df.head()
```

```python
df['source'].value_counts(normalize=True).plot(kind='bar', title='Class balance')
```

```python
df.groupby(['year', 'source']).size().unstack(fill_value=0).plot(kind='bar', figsize=(8,4), title='Headlines per year per source')
```

```python
df['len'] = df['headline'].str.len()
for src, sub in df.groupby('source'):
    plt.hist(sub['len'], bins=40, alpha=0.5, label=src)
plt.legend(); plt.xlabel('headline chars'); plt.title('Length per source')
```

```python
from collections import Counter
import re
TOK = re.compile(r"[A-Za-z']+")
for src, sub in df.groupby('source'):
    counter = Counter()
    for h in sub['headline'].astype(str): counter.update(TOK.findall(h.lower()))
    print(src, counter.most_common(20))
```
